```python
%sql
DROP VIEW IF EXISTS workspace.default.churn_scores_MOTIVES;
CREATE VIEW IF NOT EXISTS workspace.default.churn_scores_MOTIVES AS
SELECT
    *,
    RAND() AS motive_preco,
    RAND() AS motive_concorrente,
    RAND() AS motive_pagamento,
    RAND() AS motive_tecnico
FROM workspace.default.churn_scores
;
```

```python
%sql
DROP VIEW IF EXISTS workspace.default.churn_scores_MOTIVES;
CREATE VIEW IF NOT EXISTS workspace.default.churn_scores_MOTIVES AS
SELECT
    t1.*,
    t2.prob_preco AS motive_preco,
    t2.prob_concorrente AS motive_concorrente,
    t2.prob_pagamento AS motive_pagamento,
    t2.prob_tecnico AS motive_tecnico,
    t2.top_motive,
    t2.feature_digest AS motive_feature_digest,
    t2.scored_at AS motive_scored_at
FROM workspace.default.churn_scores AS t1
LEFT JOIN workspace.default.churn_motives AS t2
    ON t1.customerID = t2.customerID AND t1.source_file = t2.source_file
;
```
