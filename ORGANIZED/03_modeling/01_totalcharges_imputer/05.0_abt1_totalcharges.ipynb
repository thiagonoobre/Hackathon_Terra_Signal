# ABT 1 Creation: Imputation of `TotalCharges`

Decision to run a Linear Regression to help qith the missing values on this variable.

Check file [VariableAnalysis_for_ABT1_TotalCharges.ipynb](./VariableAnalysis_for_ABT1_TotalCharges.ipynb)  for more information on how these were selected.

```python
%sql
DROP TABLE IF EXISTS workspace.default.raw_abt1_TotalCharges;
CREATE TABLE IF NOT EXISTS workspace.default.raw_abt1_TotalCharges AS
SELECT
    customerID,
    nbrTenure,
    MonthlyCharges,
    isContractTwoYear,
    isContractOneYear,
    TotalSpend,
    TotalCharges
FROM workspace.default.feature_store_train 
WHERE nbrTenure IS NOT NULL AND TotalCharges IS NOT NULL;
```
