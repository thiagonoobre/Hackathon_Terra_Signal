# Generate new inference test file

Use random clients from `history.csv` to genereta a new input file to test the process automation of receiving a new file to score.

```python
import pandas as pd
import datetime
import os

# Read the CSV
df0 = pd.read_csv('../06_data/history.csv')
df = df0.drop(columns=['Churn'])

# Sample 15 rows randomly (without replacement)
# If your file has fewer than 15 rows, use min() to avoid errors
n = min(15, len(df))
sample = df.sample(n=n, random_state=None)  # remove random_state for true randomness

# Write out
timestamp = datetime.datetime.now().strftime("%Y%m%d_%H%M%S")
sample.to_csv(f'/Volumes/workspace/default/churn_raw/inference_{timestamp}.csv', index=False)  # volume watched by the churn-batch-scoring job
```
