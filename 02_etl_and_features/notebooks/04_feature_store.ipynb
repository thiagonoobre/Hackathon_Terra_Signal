```python
import pandas as pd

# 'train' has Churn → target_store.sql can include TargetChurn.
# New datasets don't → use target_store_inference.sql (no Churn).
datasets = {
    "train":      ["feature_store", "target_store"],
    "inference":  ["feature_store", "target_store_inference"],
}

for dataset_name, feature_names in datasets.items():
    for feature_name in feature_names:
        print(f"{feature_name} <- {dataset_name}")
        with open(f"../sql/{feature_name}.sql") as file_open:  # <-- Change path when organizeing
            query = file_open.read()

        df = spark.sql(query.format(dataset=dataset_name))

        (df.write
           .format("delta")
           .mode("overwrite")
           .saveAsTable(f"workspace.default.{feature_name}_{dataset_name}"))
```
