# ABT 2 Creation: Main Churn Model `TargetChurn`

```python
%sql
DROP TABLE IF EXISTS workspace.default.raw_abt2_Churn;
CREATE TABLE IF NOT EXISTS workspace.default.raw_abt2_Churn AS
SELECT
    t1.*,
    t2.TargetChurn
FROM workspace.default.feature_store_train AS t1
LEFT JOIN workspace.default.target_store_train AS t2
ON t1.customerID = t2.customerID
```
